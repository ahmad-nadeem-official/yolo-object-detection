```python
from ultralytics import YOLO
```

```python
def detection(path, confidence, model):
  print("welcome to obejct detection by YOLO")
  model = YOLO(model)
  result = model(source=path, show=True, conf=confidence, save=True)
  return result
```

```python
'''
the default model is yolo8n.pt
you can change it by yourself
'''

'''
| Models      | Size   | Speed (fps) | Accuracy | Use Case |
|-------------|--------|-------------|----------|----------|
| `yolov8n`   | Nano   | ⚡ Fastest   | Lower | For real-time or low-resource use |
| `yolov8s`   | Small  | Very fast   | Better   | Good balance of speed and accuracy |
| `yolov8m`   | Medium | Moderate    | High     | Balanced option |
| `yolov8l`   | Large  | Slower      | Higher   | For better accuracy |
| `yolov8x`   | X-Large| Slowest     | Best  | When accuracy is critical |
'''
```

```python
path = input("please enter the path of video :")
confidence = float(input("please enter the confidence value: "))
model = input("please enter the model :")
if model == "":
  model = "yolov8n.pt"


detection(path, confidence, model)
```
